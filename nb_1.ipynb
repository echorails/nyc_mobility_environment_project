```python
import requests
import getpass
import pandas as pd
from io import StringIO
import pyspark
```

```python
#Getting openaq data
API_KEY = getpass.getpass("API key: ")

url = "https://api.openaq.org/v3/sensors/14904963/measurements"

headers = {
    "X-API-Key": API_KEY
}
response = requests.get(url, headers=headers,params={"limit": 100, "datetime_from": "2026-01-01T00:00:00Z", "datetime_to": "2026-12-31T23:59:59Z", "page": 1}, timeout=30)
print(response.status_code)
print(response.json())
```

```python
#Getting a worldbank data
url = "https://api.worldbank.org/v2/country/USA/indicator/NY.GDP.MKTP.CD"

response = requests.get(url, params = {"format": "json", "date": "2020:2026", "per_page": 100, "page": 1}, timeout=30)
print(response.status_code)
print(response.json())
```

```python
#Getting a ECB data
url = "https://data-api.ecb.europa.eu/service/data/EXR/D.USD.EUR.SP00.A"

response = requests.get(url, params = {"format": "csvdata","startPeriod": "2026-01-01", "endPeriod": "2026-12-31" }, timeout=30)
df_ecb = pd.read_csv(StringIO(response.text))
df_ecb.head()
```
